# Calculando autovalores e autovetores

Este notebook mostra como calcular autovalores e autovetores de uma matriz de transformação. Um autovetor indica uma direção que a transformação mantém; o autovalor associado informa o fator pelo qual o vetor é escalado. Esses conceitos ajudam a compreender o comportamento de uma transformação linear e aparecem em muitas aplicações.

##### Referência: Álgebra Linear com aplicações / Anton Howard e Chris Rorres; trad. Claus Ivo Doering. - 8. ed. - Porto Alegre: Bookman, 2001.

> Se A é uma matriz n × n, um vetor não nulo **x** em $\mathbb{R}^{n}$ é um **autovetor** de A quando $A\mathbf{x}=\lambda\mathbf{x}$ para algum escalar $\lambda$. Esse escalar é chamado **autovalor** associado ao autovetor.

Em $\mathbb{R}^{2}$ e $\mathbb{R}^{3}$, a multiplicação por A envia cada autovetor para a mesma reta que passa pela origem. Se $\lambda$ for positivo, o vetor mantém o sentido; se for negativo, inverte o sentido. O valor absoluto de $\lambda$ indica o fator de escala.

```python
# importando a biblioteca numpy do Python
import numpy as np 

print("Progama que calcula Autovalores e Autovetores.\n")

# quantidade de linhas da matriz de transformação
linha = int(input("Informe a quantidade de linhas da matriz de transformação: "))

#quantidade de colunas da matriz de transformação
coluna = int(input("Informe a quantidade de colunas da matriz de transformação: "))

# inicializando uma matriz com as dimensões adequadas e de números reais
matriz = np.empty([linha, coluna], dtype = float)

# indo para a próxima linha
print("\n")

# estrutura de repetição for usada para preencher os elementos da matriz
for i in range(0, linha):
    for j in range(0, coluna):
        matriz[i][j] = float(input("Informe o elemento [{}][{}] da matriz: ".format(i + 1, j + 1)))

# imprime a matriz na tela
print("\n\nA matriz digitada é:\n\n{}".format(matriz))

if(np.linalg.det(matriz) != 0):
    
    # função responsável por receber os autovalores e autovetores da matriz de transformação
    (autovalores,autovetores) = np.linalg.eig(matriz)
    matrizDiagonal = np.diag(autovalores) # calcula a matriz diagona de autovalores
    
    # imprime na tela os autovalores
    print("\n\nAutovalores:\n\n{}\n\n".format(autovalores))

    # imprime na tela a matriz diagonal de autovalores
    print("Matriz Diagonal de autovalores:\n\n{}\n\n".format(matrizDiagonal))

    # imprime na tela os autovetores
    print("cada coluna é associada a cada um dos autovalores, respectivamente.")
    print("Autovetores:\n\n{}".format(autovetores))
else:
    print("\nA matriz não é invertível, dado que det(matriz) == 0.")
```

- Se A é uma matriz n × n e $T_A:\mathbb{R}^{n}\to\mathbb{R}^{n}$ é a transformação $T_A(\mathbf{x})=A\mathbf{x}$, então as afirmações a seguir são equivalentes:

|**Afirmações equivalentes**|
|:--------------------------|
|1. A é invertível.|
|2. $A\mathbf{x}=0$ tem somente a solução trivial.|
|3. A forma escalonada reduzida por linhas de A é $I_n$.|
|4. A pode ser escrita como produto de matrizes elementares.|
|5. $A\mathbf{x}=\mathbf{b}$ é consistente para todo vetor $\mathbf{b}\in\mathbb{R}^{n}$.|
|6. $A\mathbf{x}=\mathbf{b}$ tem exatamente uma solução para todo vetor $\mathbf{b}\in\mathbb{R}^{n}$.|
|7. $\det(A)\ne0$.|
|8. A imagem de $T_A$ é $\mathbb{R}^{n}$.|
|9. $T_A$ é injetora.|
|10. As colunas de A são linearmente independentes.|
|11. As linhas de A são linearmente independentes.|
|12. As colunas de A geram $\mathbb{R}^{n}$.|
|13. As linhas de A geram $\mathbb{R}^{n}$.|
|14. As colunas de A formam uma base de $\mathbb{R}^{n}$.|
|15. As linhas de A formam uma base de $\mathbb{R}^{n}$.|
|16. A tem posto n.|
|17. A tem nulidade 0.|
|18. O complemento ortogonal do espaço nulo de A é $\mathbb{R}^{n}$.|
|19. O complemento ortogonal do espaço linha de A é $\{0\}$.|
|20. $A^{T}A$ é invertível.|
|21. $\lambda=0$ não é autovalor de A.|
