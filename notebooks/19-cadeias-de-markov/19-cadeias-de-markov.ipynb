# Cadeias de Markov

Uma cadeia de Markov modela um sistema que muda entre estados. Sua propriedade principal é que, conhecendo o estado atual, a distribuição do próximo estado não depende dos estados anteriores.

##### Álgebra Linear com aplicações / Anton Howard e Chris Rorres; trad. Claus Ivo Doering. - 8. ed. - Porto Alegre: Bookman, 2001.

### Um Processo de Markov

Considere um sistema que, em cada observação, pode estar em um de um número finito de estados. Por exemplo, o tempo pode ser ensolarado, nublado ou chuvoso. Se a probabilidade do próximo estado depende somente do estado atual, o processo é uma cadeia de Markov.

### Definição

Denotamos por 1, 2, ..., k os k estados possíveis. A probabilidade de passar do estado j para o estado i é $p_{ij}$. A matriz $P=[p_{ij}]$ é a matriz de transição. Nesta convenção, as colunas representam o estado atual e as linhas representam o próximo estado.

Para uma cadeia de Markov com três estados, a matriz de transição tem a forma:

$p_{ij}$: linhas indicam o próximo estado i e colunas indicam o estado atual j.

$$P=\begin{bmatrix}
p_{11}&p_{12}&p_{13}\\
p_{21}&p_{22}&p_{23}\\
p_{31}&p_{32}&p_{33}
\end{bmatrix}$$

Por exemplo, $p_{32}$ é a probabilidade de passar do estado 2 para o estado 3; $p_{11}$ é a probabilidade de permanecer no estado 1.

#### Exemplo: locadora de automóveis

Uma locadora tem três lojas, identificadas por 1, 2 e 3. Um carro pode ser devolvido em qualquer uma delas. O gerente estima as probabilidades de devolução mostradas a seguir:

```python
# definindo a matriz de transição A
A = np.array([[80/100, 30/100, 20/100],
              [10/100, 20/100, 60/100],
              [10/100, 50/100, 20/100]])
```

```python
# imprimindo a matriz aluguelCarro
print("A matriz de transição A é:\n\n{}".format(A))
```

Essa é a matriz de transição do modelo. Por exemplo, um carro alugado na loja 3 tem probabilidade 0,6 de ser devolvido na loja 2; um carro alugado na loja 1 tem probabilidade 0,8 de voltar à loja 1.

#### Exemplo: doações à universidade

A secretaria de uma associação de ex-alunos observa que 80% dos ex-alunos que doam em um ano também doam no ano seguinte. Entre os que não doam, 30% passam a doar no ano seguinte. Podemos modelar isso com dois estados: “doa” e “não doa”. A matriz de transição é:

```python
# definindo a matriz de transição P
P = np.array([[80/100, 30/100],
              [20/100, 70/100]])
```

```python
# imprimindo a matriz P
print("A matriz de transição P é: \n\n{}".format(P))
```

Nas matrizes de transição deste capítulo, as entradas de cada coluna somam 1. Se o sistema está no estado j, ele certamente estará em um dos k estados possíveis na observação seguinte. Portanto,
$$p_{1j}+p_{2j}+\cdots+p_{kj}=1.$$

Uma matriz cujas colunas somam 1 e cujas entradas são não negativas é chamada **matriz estocástica** ou **matriz de probabilidade**. Toda matriz de transição, nesta convenção, é estocástica.

Em geral, não sabemos com certeza em qual estado o sistema estará. Podemos descrever as probabilidades dos estados por um vetor-coluna. Para três estados, escrevemos:

$$x = \begin{bmatrix}
              x_{1} \\
              x_{2} \\
              x_{3} \\
              \end{bmatrix}$$

O componente $x_i$ representa a probabilidade de o sistema estar no estado i naquela observação. Em geral:

> O **vetor-estado** de uma cadeia de Markov com k estados é um vetor-coluna cuja i-ésima entrada é a probabilidade de o sistema estar no estado i.

As entradas do vetor-estado são não negativas e somam 1. Um vetor com essas propriedades é chamado **vetor de probabilidade**.

Se conhecemos o vetor-estado inicial $x^{(0)}$, podemos calcular os vetores das observações seguintes:

> **Teorema:** se P é a matriz de transição e $x^{(n)}$ é o vetor-estado na observação n, então $x^{(n+1)}=Px^{(n)}$.

A demonstração usa resultados de probabilidade. Aplicando o teorema repetidamente, obtemos:

| |
|--|
|$x^{(1)}=Px^{(0)}$|
|$x^{(2)}=Px^{(1)}=P^{2}x^{(0)}$|
|$x^{(3)}=Px^{(2)}=P^{3}x^{(0)}$|
|$\vdots$|
|$x^{(n)}=Px^{(n-1)}=P^{n}x^{(0)}$|

Assim, o vetor inicial $x^{(0)}$ e a matriz P determinam o estado probabilístico em cada observação posterior.

#### Exemplo: doações à universidade

A matriz de transição do exemplo das doações é:

```python
# imprimindo a matriz P
print("A matriz de transição P: \n\n{}".format(P))
```

Vamos acompanhar um recém-formado que não fez uma doação no primeiro ano após a formatura. Inicialmente, o sistema está no estado 2, então o vetor-estado inicial é:

```python
# definindo o vetor-estado x
x = np.array([[0],
              [1]])
```

```python
# imprimindo o vetor-estado x
print("O vetor-estado é: \n\n{}".format(x))
```

Pelo teorema, o vetor-estado após uma observação é:

```python
# função que realizará a mudança de estado até o ano a ser definido
def estado(matrizTransicao, vetorEstado, transicao):
    
    # estrutura de repetição for de 0 até (anos - 1)
    for indice in  range(0, transicao):
        
        # realiza o calculo definido no teorema acima
        vetorEstado = np.dot(matrizTransicao, vetorEstado)
    
    # retorna o vetor estado após sofrer todas as mudanças de estado até o ano informado
    return vetorEstado
```

```python
# vetor-estado após 1 ano
print("O vetor-estado após 1 ano é: \n\n{}".format(estado(P, x, transicao = 1)))
```

Após um ano, o modelo atribui 30% de probabilidade ao estado “doa” e 70% ao estado “não doa”.

```python
# vetor-estado após 8 ano
print("O vetor-estado após 8 anos é: \n\n{}".format(estado(P, x, transicao = 8)))
```

```python
# vetor-estado após 11 ano
print("O vetor-estado após 11 ano é: \n\n{}".format(estado(P, x, transicao = 11)))
```

Neste exemplo, os vetores-estado após 8 e 11 anos são próximos. A distribuição de probabilidades se estabiliza à medida que o número de observações aumenta.

#### Exemplo: locadora de automóveis

Vamos retomar o exemplo da locadora.

```python
# imprimindo a matriz de transição do exemplo 1
print("A matriz de transição A do exemplo 1 é: \n\n{}".format(A))
```

```python
# definindo o vetor-estado inicial
w = np.array([[0],
              [1],
              [0]])
```

```python
# imprimindo o vetor w
print("O vetor-estado w é: \n\n{}".format(w))
```

Esse vetor representa a certeza de que o aluguel começou na loja 2.

```python
# vetor-estado após a 2º observação
print("O vetor-estado na 2º observação é: \n\n{}".format(estado(A, w, transicao = 2)))
```

E, após a quarta observação:

```python
# vetor-estado após a 4º observação
print("O vetor-estado na 4º observação é: \n\n{}".format(estado(A, w, transicao = 4)))
```

Na quarta observação, as probabilidades são aproximadamente 51% para a loja 1, 26% para a loja 2 e 22% para a loja 3. A soma pode diferir ligeiramente de 100% por causa do arredondamento.

```python
# vetor-estado após a 11º observação
print("O vetor-estado na 11º observação é: \n\n{}".format(estado(A, w, transicao = 11)))
```

> Neste exemplo, não precisamos modelar quanto tempo o cliente permaneceu com o carro: cada passo corresponde a uma devolução. As probabilidades se aproximam de uma distribuição estável à medida que o número de observações cresce.

#### Exemplo: guarda de trânsito

Uma guarda de trânsito controla oito cruzamentos de uma rua. A cada hora, ela permanece no cruzamento atual ou segue para um cruzamento adjacente. Entre as opções permitidas, cada uma é escolhida com a mesma probabilidade. Por exemplo, se está no cruzamento 5, seu próximo cruzamento pode ser 2, 4, 5 ou 8, cada um com probabilidade $1/4$. No dia seguinte, ela começa no cruzamento em que terminou.

```python
# definindo a matriz de transição C
C = np.array([[1/3, 1/3, 0  , 1/5, 0  , 0  , 0  , 0  ],
              [1/3, 1/3, 0  , 0  , 1/4, 0  , 0  , 0  ],
              [0  , 0  , 1/3, 1/5, 0  , 1/3, 0  , 0  ],
              [1/3, 0  , 1/3, 1/5, 1/4, 0  , 1/4, 0  ],
              [0  , 1/3, 0  , 1/5, 1/4, 0  , 0  , 1/3],
              [0  , 0  , 1/3, 0  , 0  , 1/3, 1/4, 0  ],
              [0  , 0  , 0  , 1/5, 0  , 1/3, 1/4, 1/3],
              [0  , 0  , 0  , 0  , 1/4, 0  , 1/4, 1/3]])
```

```python
# imprimindo a matric C
print("A matriz de transição C é: \n\n{}".format(C))
```

A entrada da matriz na linha i e coluna j representa a probabilidade de a guarda ir do cruzamento atual j para o cruzamento i.

Se ela começar no cruzamento 5, os vetores abaixo descrevem as probabilidades de sua localização a cada hora:

```python
# determinando o vetor-estado inicial y se ela começar no cruzamento 5
y = np.array([[0],
              [0],
              [0],
              [0],
              [1],
              [0],
              [0],
              [0]])
```

```python
# imprimindo o vetor-estado y
print("O vetor-estado y é: \n\n{}".format(y))
```

```python
# vetor-estado após a 1º observação
print("O vetor-estado na 1º observação é: \n\n{}".format(estado(C, y, transicao = 1)))
```

A partir do cruzamento 5, a guarda não pode ir diretamente aos cruzamentos 1, 3, 6 ou 7. Cada um dos cruzamentos 2, 4, 5 e 8 tem probabilidade $1/4$ de ser escolhido.

```python
# vetor-estado após a 10º observação
print("O vetor-estado na 10º observação é: \n\n{}".format(estado(C, y, transicao = 10)))
```

```python
# vetor-estado após a 22º observação
print("O vetor-estado na 22º observação é: \n\n{}".format(estado(C, y, transicao = 22)))
```

Neste exemplo, para n maior que 22, as entradas dos vetores-estado coincidem com as de $x^{(22)}$ até três casas decimais. A distribuição se estabiliza conforme aumentam as observações.
