# Sistemas lineares

Um sistema linear é um conjunto de equações lineares nas mesmas incógnitas. Sua solução deve satisfazer todas as equações do sistema.

##### Referência: Álgebra Linear com aplicações / Anton Howard e Chris Rorres; trad. Claus Ivo Doering. - 8. ed. - Porto Alegre: Bookman, 2001.

### Sistemas Lineares:

Um conjunto finito de equações lineares nas variáveis $x_{1}, x_{2}, ..., x_{n}$ é chamado sistema linear. Uma sequência $s_{1}, s_{2}, ..., s_{n}$ é solução quando a substituição $x_{1}=s_{1}, x_{2}=s_{2}, ..., x_{n}=s_{n}$ satisfaz cada equação.

Um sistema sem solução é **inconsistente**. Se tiver ao menos uma solução, é **consistente**.

> Um sistema linear tem nenhuma solução, uma única solução ou infinitas soluções.

### Matrizes Aumentadas

Ao organizar os coeficientes e os termos constantes, podemos representar um sistema com m equações e n incógnitas por uma matriz aumentada:

$\begin{bmatrix}
          a_{11} & a_{12} & ... & a_{1n} & b_{1} \\
          a_{21} & a_{22} & ... & a_{2n} & b_{2} \\
            :    &   :    &  :  &    :   &   :   \\
          a_{m1} & a_{m2} & ... & a_{mn} & b_{m} \\
          \end{bmatrix}$

Essa representação é chamada **matriz aumentada** do sistema.

### Exemplo: Resolvendo Sistemas Lineares com Python

Considere o seguinte sistema linear:


- a + b + c + d + e + f = 21
- 10a + 2b + 3c + 4d +5e + f = 70
- a + b + 2c + 4d + e + f = 36
- 2a + 7b + c + d + 10e + 2f = 85 
- 3a + b + c + d + 20e + f = 118
- 2a + 6b + c + 4d + 5e + f = 64

Resolver esse sistema manualmente pode ser trabalhoso. Como a matriz de coeficientes deste exemplo é invertível, podemos usar a relação $X=A^{-1}B$ para encontrar a solução. Em outros sistemas, a matriz pode não ser invertível; nesse caso, é necessário usar outro método.

Podemos representar o sistema linear acima utilizando matrizes:

$$\begin{bmatrix}
         1 & 1 & 1 & 1 & 1 & 1\\
         10 & 2 & 3 & 4 & 5 & 1\\
         1 & 1 & 2 & 4 & 1 & 1\\
         2 & 7 & 1 & 1 & 10 & 2\\
         3 & 1 & 1 & 1 & 20 & 1\\
         2 & 6 & 1 & 4 & 5 & 1\\
         \end{bmatrix}.\begin{bmatrix}
                               a_{1} \\
                               b_{1} \\
                               c_{1} \\
                               d_{1} \\
                               e_{1} \\
                               f_{1} \\
                               \end{bmatrix} = \begin{bmatrix}
                                                 21 \\
                                                 70 \\
                                                 36 \\
                                                 85 \\
                                                 118 \\
                                                 64 \\
                                                 \end{bmatrix}$$

Vamos seguir os passos abaixo para resolver o sistema. O método da inversa se aplica quando a matriz quadrada de coeficientes é invertível.

$$AX = B$$

$$A.A^{-1}.X = A^{-1}.B$$

$$X = A^{-1}.B$$

```python
# importando a biblioteca numpy
import numpy as np 
```

```python
# declarar as matrizes do exemplo acima
A = np.array([[1,1,1,1,1,1],[10,2,3,4,5,1],[1,1,2,4,1,1],[2,7,1,1,10,2],
              [3,1,1,1,20,1],[2,6,1,4,5,1]])
B = np.array([[21],[70],[36],[85],[118],[64]])

# Encontrar a inversa de A
A_I = np.linalg.inv(A) #Função que encontra a matriz inversa

# Encontrando X
X = np.dot(A_I,B) # Função que realiza mutiplicação entre matrizes sem especificador de tipo
# X = np.linalg.solve(A,B) #Função alternativa que resolve Sistemas Lineares
```

```python
# imprime a matriz A dos coeficientes de X e Y
print("{}".format(A)) 
```

```python
# imprime a matriz A inversa
print("{}".format(A_I)) 
```

```python
# imprime a matriz B resultado dos sistemas lineares
print("{}".format(B)) 
```

```python
# imprime a matrix , o  conjunto solução desse sistema
print("{}".format(X)) 
```
